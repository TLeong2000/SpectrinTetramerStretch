# Changelog

## 4/13/2024

It turns out that during my 4/6 update, I forgot to update the `totLength` parameter to match the corrected change of the total length of the polymer - which explains the unphysical looking trajectories for displacement ratios greater than 0.5 - because the `totLength` parameter was still the old value, which is twice the corrected value of the total length of the polymer, a displacement ratio greater than 0.5 effectively corresponded to enthalpic stretching of the polymer, which is not what we desire to investigate.

I updated the `totLength` parameter back to the corrected value of the total length of the polymer. I also halved the timestep size from the previous value of $100\, \text{fs}$ to $50\, \text{fs}$ and doubled the number of timesteps the dynamics ran for from $10E6$ to $20E6$ to keep the total physical simulation time the same. I chose to halve the timestep size because I noticed that even for displacement ratios less than 0.5 on the 4/6 data (which correspond to entropic stretching), the magnitude of the force still severely diverged from the analytical predictions - which could have been a sign of physical instability of the simuation.

After plotting the data, it appears that there is a local maxima in the observed force of the 84 bead model at a displacement ratio of around 0.35 - which is behavior that does not follow the BT model. Given that the force still does not smoothly go to zero as displacement ratio goes to zero, this could be due to discrete sampling effects.

More concerningly, the zero-crossing point for the observed force of the 84 bead model takes place at a displacement ratio that is further away from the BT model's analytical prediction compared to the 42 bead model:



I am not sure what could be the cause of this. Subsequently, I created a new script in order to visualize the trajectories of the chain at a displacement ratio of 0.97 (which is at the observed zero-crossing point for the 84 bead model) and 0.90 (which is the analytically-predicted zero-crossing point) to find out if there is anything obviously unphysical with the trajectories.

I confirmed with Dr. Peng that the bond coefficient is analytically irrelevant for displacment ratios that fall within the scope of the Blundell-Terentjev model. However, after taking a look at the trajectory visualization for the displacement ratio of 0.90, there appears to be some very high frequency bond vibrational motion that I think is responsible for why the force is not zero. Whether that bond vibrational motion is an artifact of too high of a timestep or if it is genuinely physical, I do not know.

## 4/6/2024

The persistence length is defined as:

$$l_p := \frac{k_a * l_0}{k_B * T}$$

where $k_a$ is the bending angle coefficient, $l_0$ is the equilibrium bond length, $k_B$ is Boltzmann's constant, and $T$ is the temperature

The 42 bead model that I had featured...

a bending angle coefficient of:

$$k_a = 64.717\, \frac{\text{kcal}}{\text{mol}}$$

and an equilibrium bond length of:

$$l_0 = 27.350 \, \mathring{\text{A}}$$

The simulation was run at a temperature of $310\, \text{K}$. This means that for dimensional consistency, Boltzmann's constant must be:

$$k_B = 1.987 * 10^{-3}\, \frac{\text{kcal}}{\text{mol}*\text{K}}$$

As a result, our persistence length must be:

$$l_p = \frac{64.717 * 27.350}{1.987 * 10^{-3} * 310}\, \mathring{\text{A}}$$
$$l_p = 2873.533\, \mathring{\text{A}}$$

Now for the 84 bead model, when I originally constructed the model, I completely forgot that the purpose of it was to be a higher-resolution refinement of the 42 bead model - I did not shorten the equilibrium bond lengths to keep the total equilibrium length of the polymer constant. I shall rectify that now.

The 42 bead model, being linear, has 41 bonds, and the 84 bead model, also being linear, has 83 bonds. Therefore, the equilibrium length an individual bond in the 84 bead model must be equal to the equilibrium length of a bond in the 42 bead model times $\frac{41}{83}$:

$$l_{\text{eq}\, 84} = 27.350 \, \mathring{\text{A}} * \frac{41}{83} = 13.510 \, \mathring{\text{A}} $$

However, to keep the persistence lengths of the 84 and 42 bead models equal to each other, we need to multiply the bending angle coefficient by $\frac{83}{41}$:

$$k_{a\, 84} = 64.717\, \frac{\text{kcal}}{\text{mol}} * \frac{83}{41} = 131.013\, \frac{\text{kcal}}{\text{mol}}$$

Obviously, Boltzmann's constant does not change, and we will still be using the same temperature of $310 \, \text{K}$.

The below table summarizes the data file for the 42 vs 84 bead model:

| Num Beads                                                                          | 42     | 84      |
|------------------------------------------------------------------------------------|--------|---------|
| Bond Coefficient $\left(\frac{\text{kcal}}{\text{mol}*\mathring{\text{A}}}\right)$ | 3.443  | 3.443   |
| Bond Equilibrium Length $\left(\mathring{\text{A}}\right)$                         | 27.350 | 13.510  |
| Angle Coefficient $\left(\frac{\text{kcal}}{\text{mol}}\right)$                    | 64.717 | 131.013 |
| Angle Equilibrium Displacement $\left(\degree\right)$                              | 180    | 180     |

I created a new prototype file for the 84 bead model, ran a relaxation on it, recentered the center-of-mass to be at $(0,0,0)$, and will run simulations for it.

After plotting the force data, it appears to be utterly hideous - the force crosses the zero point at a displcement ratio of ~0.45 and linearly rises:



This is not surprising, given that we increased the angle coefficient but kept the bond coefficient the same. I am not sure how the bond coefficient plays into the Blundell-Terentjev model, which is why I haven't changed it. That said, I do not think that increased sampling will rectify this issue - I suspect that the bond coefficient comes into play here, but I don't know why.
